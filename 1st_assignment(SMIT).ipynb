### Q3: Arithmetic Operator

```python
num_1 = 10
num_2 = 20

print("additon")
print(num_1 + num_2)
print("subtraction")
print(num_1 - num_2)
print("multiplication")
print(num_1 * num_2)
print("division")
print(num_1 / num_2)
print("power")
print(num_1 ** num_2)
print("modulas")
print(num_1 % num_2)
```

### Q6: Rectangle Area

```python
length = int(input("enter length: "))
width = int(input("enter width: "))

print(f"this is area: {length*width}")
print(f"this is perimeter: {2*(length+width)}")
```

### Q8: Student Eligibility

```python
mark = int(input("enter your marks: "))
attendence = int(input("enter your attendence percentage: "))

if mark >= 50 and attendence >= 75:
    print("you are eligible")
else:
    print("you are not eligible")
```
