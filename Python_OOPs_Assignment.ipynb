**Q2:** Aik `BankAccount` class banayein jismein `__balance` private variable ho. Aik `deposit()` aur aik `withdraw()` method banayein, check karein ke withdraw karte waqt balance zero se kam na ho.

```python
# Write your code here for Q2
class bankaccount:
    def __init__(self):
        self.__balance = 100

    def deposit(self, deposit_amount):
        self.__balance += deposit_amount
        print("current balance after deposit", self.__balance)

    def withdraw(self, withdraw_amount):
        if withdraw_amount > 0:
            self.__balance -= withdraw_amount
            print("this is your current balance after withdraw", self.__balance)
        else:
            print("insufition amount")
a1 = bankaccount()
a1.deposit(100)
a1.withdraw(50)
```

**Q10:** Hierarchical Inheritance kya hoti hai? Aik parent class `Shape` banayein aur us se do alag child classes `Circle` aur `Square` banakar samjhayein.

```python
# Write your code here for Q10
class shape:
    def __init__(self,num1):
        self.num1 = num1

class circle(shape):
    def __init__(self, num1):
        super().__init__(num1)

    def circle_value(self):
        circle = round((self.num1 * self.num1)*3.1416,2)
        print("circle value", circle)

class square(shape):
    def __init__(self, num1):
        super().__init__(num1)

    def square(self):
        squared = self.num1**2
        print("square value",squared)
        
c1 = circle(5)
c1.circle_value()
s1 = square(4)
s1.square()
```

**Q13:** Operator Overloading kya hoti hai? Do objects ko `+` operator ke zariye add karne ke liye `__add__` magic method ka use kaise karte hain? Code likhein.

```python
# Write your code here for Q13
class Number:
    def __init__(self, value):
        self.value = value

    def __add__(self, other):
        return Number(self.value + other.value)

    def __str__(self):
        return str(self.value)

n1 = Number(10)
n2 = Number(20)

n3 = n1 + n2   
print(n3)
```

**Q15:** Kya Python default roop se Method Overloading (same method name with different arguments) support karta hai? Agar nahi, toh hum same name ke method ko multiple arguments ke sath kaise handle karte hain? Code likhein.

```python
# Write your code here for Q15
class Calculator:
    def add(self, *args):
        total = 0
        for num in args:
            total += num
        return total

calc = Calculator()
print(calc.add(5))           
print(calc.add(5, 10))       
print(calc.add(1, 2, 3, 4))  
```

**Q20:** Aik real-world scenario (jaise Payment System) design karein jahan `Payment` aik abstract class ho aur `CreditCardPayment` aur `UPIPayment` uski subclasses hon.

```python
# Write your code here for Q20
class Payment(ABC):
    def __init__(self, amount):
        self.__amount = amount  

    def get_amount(self):
        return self.__amount

    @abstractmethod
    def pay(self):
        pass

class CreditCardPayment(Payment):
    def pay(self):
        print(f"Paid {self.get_amount()} using Credit Card")

class UPIPayment(Payment):
    def pay(self):
        print(f"Paid {self.get_amount()} using UPI")

payments = [CreditCardPayment(5000), UPIPayment(1200)]
for p in payments:
    p.pay()
```
